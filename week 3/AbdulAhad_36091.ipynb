```python
import pandas as pd

data = pd.read_csv(r"C:\Users\Ahad\Downloads\Assignment_1_Completed_AI_Assisted (1)\assignments\week1\teaching_prices_clean.csv")

data['date'] = pd.to_datetime(data['date'])

print("Rows:", len(data))
print("Items:", data['item'].nunique())
print("Dates:", data['date'].nunique())
print("Cities:", data['city'].nunique())

first_date = data['date'].min()
last_date = data['date'].max()

for item in data['item'].unique():
    item_data = data[data['item'] == item]
    mean_val = item_data['price'].mean()
    median_val = item_data['price'].median()
    
    first_prices = item_data[item_data['date'] == first_date]['price']
    last_prices = item_data[item_data['date'] == last_date]['price']
    
    f_mean = first_prices.mean()
    l_mean = last_prices.mean()
    change = l_mean - f_mean
    percent = (change / f_mean) * 100
    
    print("\nItem:", item)
    print("Mean:", round(mean_val, 2))
    print("Median:", round(median_val, 2))
    print("First date mean:", round(f_mean, 2))
    print("Last date mean:", round(l_mean, 2))
    print("Percentage change:", round(percent, 2), "%")
```
