```python
import pandas as pd
```

```python
shop_df = pd.read_csv(r"C:\Users\Ahad\Downloads\pakistani_ecommerce_dataset.csv")
```

```python
shop_df.head()
```

```python
shop_df.columns
```

Q1: Which is the most popular product category?

```python
shop_df["product_category"].value_counts()
```

Ans: home decor

Q2: What is the average order value?

```python
shop_df["order_value"].mean()
```

Q3: which city has highest number of orders?

```python
shop_df["city"].value_counts()
```

Karachi

Q4: What is the most used payment method?

```python
shop_df["payment_method"].value_counts()
```

cash on delivery

Q5: Draw a bar chart showing average order value by city.

```python
import matplotlib.pyplot as plt
```

```python
city_mean_val = shop_df.groupby("city")["order_value"].mean()
city_mean_val
```

```python
plt.bar(city_mean_val.index, city_mean_val.values, color="teal", width=0.45)
plt.title("Average Order Value Across Cities")
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.show()
```

Q1: What is the average number of study hours?

```python
habits_df["study_hours"].mean()
```

4.40 HOURS

Q2: Who sleeps more on average - males or females?

```python
habits_df.groupby("gender")["sleep_hours"].mean()
```

Females get more sleep

Q3: Is there a relationship between screen time and sleep hours?

```python
habits_df["screen_time"].corr(habits_df["sleep_hours"])
```

0.052, no linear RS between screen time and sleep

Q4: Draw a histogram of daily transport cost

```python
plt.hist(habits_df["daily_transport_cost"], bins=8, color="gold", edgecolor="black")
plt.title("Daily Transport Cost Distribution")
plt.xlabel("Daily Transport Cost")
plt.ylabel("Frequency")
plt.show()
```

Q5: Compare average internet usage by gender.

```python
net_by_gender = habits_df.groupby("gender")["internet_usage"].mean()
net_by_gender
```

```python
plt.bar(net_by_gender.index, net_by_gender.values, color="slateblue", width=0.45)
plt.title("Average Internet Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Internet Usage")
plt.show()
```
