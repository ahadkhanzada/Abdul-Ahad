```python
import pandas as pd
```

```python
data = pd.read_csv("teaching_prices_clean.csv")
print("CSV loaded.")
```

```python
display(data.head())
print(data.columns.tolist())
```

 dataset shape

```python
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
print("Shape:", data.shape)
```

 distinct items, dates and cities


```python
data["date"] = pd.to_datetime(data["date"])

print("Total rows:", len(data))
print("Different items:", data["item"].nunique())
print("Different dates:", data["date"].nunique())
print("Different cities:", data["city"].nunique())
print("Missing values:", data.isnull().sum().sum())
```

 average item prices on the first and last dates


```python
first_date = data["date"].min()
last_date = data["date"].max()

print("First date:", first_date.date())
print("Last date:", last_date.date())

for item in data["item"].unique():

    first_price = data[
        (data["item"] == item) &
        (data["date"] == first_date)
    ]["price"].mean()

    last_price = data[
        (data["item"] == item) &
        (data["date"] == last_date)
    ]["price"].mean()

    change = last_price - first_price

    percentage = (change / first_price) * 100

    print("\nItem:", item)
    print("First average:", round(first_price, 2))
    print("Last average:", round(last_price, 2))
    print("Price change:", round(change, 2))
    print("Percentage change:", round(percentage, 2), "%")
```

Comparision City 1 and City 2


```python
for item in data["item"].unique():

    city1 = data[
        (data["item"] == item) &
        (data["city"] == "City 1")
    ]["price"].mean()

    city2 = data[
        (data["item"] == item) &
        (data["city"] == "City 2")
    ]["price"].mean()

    difference = city2 - city1

    print("\nItem:", item)
    print("City 1 average:", round(city1, 2))
    print("City 2 average:", round(city2, 2))
    print("Difference:", round(difference, 2))
```

 Interpretation

The first-to-last comparison shows an increase for all four item averages in this dataset. Sweetener D has the largest percentage increase (about 19.41%), and Pulse C has the smallest (about 3.69%). City 2 is PKR 6 per kg higher than City 1 for every matching item and date.

**Limitation:** The data label explicitly says these are illustrative teaching values, not actual PBS observations. The dataset does not include household income, wages, spending, transport costs, supply conditions or reasons for price changes. It cannot establish affordability for households or explain why prices changed.
