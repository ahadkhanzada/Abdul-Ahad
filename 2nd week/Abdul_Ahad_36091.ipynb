```python
# ASSIGNMENT 2 THE DIRTY LEDGER - COMPLETE CLEANING CODE


import pandas as pd
import numpy as np


data = pd.read_csv(r"C:\Users\Ahad\Downloads\mids + finals ica prep\data\teaching_prices_messy.csv")
print("Initial Dataset Check:")
print("Rows:", len(data))
print("Items:", data["item"].nunique())
print("Dates:", data["date"].nunique())
print("Cities:", data["city"].nunique())
print("Missing values:", data["price"].isna().sum())
print("Duplicates:", data.duplicated().sum())


data_clean = data.copy()


data_clean = data_clean.drop_duplicates().reset_index(drop=True)


data_clean.loc[(data_clean['date'] == '2026-02-05') & (data_clean['item'] == 'oil b'), 'item'] = 'Staple A'


def fix_price(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).replace(',', '').strip()
    num = float(val_str)
    if num == 9999.0:
        return np.nan
    if num == 1250.0:
        return 109.0
    return num

data_clean['price'] = data_clean['price'].apply(fix_price)


data_clean.loc[(data_clean['date'] == '2026-01-01') & (data_clean['item'] == 'Pulse C') & (data_clean['city'] == 'City 2'), 'price'] = 206.0


data_clean.loc[(data_clean['date'] == '2026-01-22') & (data_clean['item'] == 'Sweetener D') & (data_clean['city'] == 'City 1'), 'price'] = 157.5


data_clean["date"] = pd.to_datetime(data_clean["date"])

print("\nAfter Cleaning Check:")
print("Rows:", len(data_clean))
print("Missing values:", data_clean["price"].isna().sum())


data_clean.to_csv("teaching_prices_clean.csv", index=False)
print("Cleaned file saved successfully as teaching_prices_clean.csv")


first_date = data_clean["date"].min()
last_date = data_clean["date"].max()

print("\nFirst date:", first_date.date())
print("Last date:", last_date.date())

for item in data_clean["item"].unique():
    first_price = data_clean[(data_clean["item"] == item) & (data_clean["date"] == first_date)]["price"].mean()
    last_price = data_clean[(data_clean["item"] == item) & (data_clean["date"] == last_date)]["price"].mean()
    change = last_price - first_price
    percent = (change / first_price) * 100
    print("\nItem:", item)
    print("First average:", round(first_price, 2))
    print("Last average:", round(last_price, 2))
    print("Change:", round(change, 2))
    print("Percentage change:", round(percent, 2), "%")


print("\nChecking City Differences (City 2 vs City 1):")
for item in data_clean["item"].unique():
    differences = []
    for date in data_clean["date"].unique():
        city1 = data_clean[(data_clean["item"] == item) & (data_clean["date"] == date) & (data_clean["city"] == "City 1")]["price"].iloc[0]
        city2 = data_clean[(data_clean["item"] == item) & (data_clean["date"] == date) & (data_clean["city"] == "City 2")]["price"].iloc[0]
        differences.append(city2 - city1)
    print(item, "price differences:", set(differences))
```
